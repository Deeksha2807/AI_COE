```python

import tensorflow as tf
from tensorflow.keras import layers, Model

X = tf.constant([[1, 2], [2, 3], [3, 4], [1, 3]])
y = tf.constant([3, 4, 1, 2])


inputs = layers.Input(shape=(2,))
x = layers.Embedding(5, 8)(inputs)

a = layers.MultiHeadAttention(
    num_heads=2, key_dim=4
)(x, x)

x = layers.GlobalAveragePooling1D()(a)
outputs = layers.Dense(5, activation="softmax")(x)

model = Model(inputs, outputs)
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.fit(X, y, epochs=5, verbose=1)
prediction = model.predict(
    tf.constant([[1, 2]]), verbose=0
)
print("Predicted next token:", prediction.argmax())
model.save("transformer.keras")



```
